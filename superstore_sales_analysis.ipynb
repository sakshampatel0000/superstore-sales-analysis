## sales by category analysis

```python
#This section analyzes the total sales for each product category using a bar chart.
import matplotlib.pyplot as plt 

category_sales = Data.groupby("Category")["Sales"].sum()    # data.groupby("category") kept same ctegoryb together 


plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)  # When groupby() is applied, Pandas automatically uses the grouped column as the index 

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python

## Insights

#- Technology generated the highest sales.
#- Furniture generated comparatively lower sales.
#- Office Supplies performed moderately.
```

## Profit by Category Analysis



```python
#This section analyzes the total profit generated by each product category using a bar chart.

profit=Data.groupby("Category")["Profit"].sum()

plt.bar(profit.index , profit.values)

plt.title("profit by category")
plt.xlabel("category")
plt.ylabel("profit")

plt.show()


```

```python
 ## Insights

# Technology generated the highest profit.
# Office Supplies generated a moderate profit.
# Furniture generated the lowest profit.
```

## Sales by Sub_Category Analysis

```python
Sub_Category= Data.groupby("Sub-Category")["Sales"].sum()

plt.figure(figsize=(14,6))
plt.bar(Sub_Category.index , Sub_Category.values)

plt.title("Sales by Sub_Category")
plt.xlabel("Sub_category")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.show()
```

```python
 ## insights 

#Top-performing chairs  generated the highest sales and contribute a major portion of the business revenue.
#Phones and Chairs  are among the best selling sub-categories.
#Some sub-categories show very low sales, indicating lower customer demand or limited product performance.
#There is a large variation in sales across different sub-categories, which means customer demand is not evenly distributed.
#The company should focus on high-selling sub-categories while analyzing why low-selling ones are underperforming.

```

## Sales vs profit analysis 

```python
# Sales vs Profit Analysis

#This scatter plot shows the relationship between Sales and Profit.

#Goal:
#- Check whether higher sales always generate higher profit.
#- Identify loss-making orders despite high sales.

plt.figure(figsize=(8,6))

plt.scatter(Data["Sales"], Data["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

## Sales distribution

```python
# Sales Distribution

#This histogram shows how sales values are distributed across all orders.

#Goal:
#- Understand the frequency of sales values.
#- Identify whether most orders have low or high sales.
#- Detect outliers in sales.



plt.figure(figsize=(8,5))

plt.hist(Data["Sales"], bins=30)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
## Insights

#- Most orders have relatively low sales.
#- Only a few orders generated very high sales.
#- Sales distribution is right-skewed.
#- High-value sales orders are comparatively rare.
```

## Profit distribution 

```python
# Profit Distribution

#This histogram shows the distribution of profit values.

#Goal:
#- Analyze how profit is spread across orders.
#- Identify profit concentration.
# Observe loss-making orders.


plt.figure(figsize=(8,5))

plt.hist(Data["Profit"], bins=30)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
## Insights

#- Most orders generated small profits.
#- Some orders resulted in losses.
#- A few orders generated very high profit.
#- Profit distribution is uneven across orders.
```

## corelation heatmap

```python
# Correlation Heatmap

#This heatmap shows the correlation between numerical features in the dataset.
#It helps identify positive, negative, or weak relationships among Sales, Profit, Quantity, and Discount.

import seaborn as sns

plt.figure(figsize=(6,4))

sns.heatmap(
    Data[["Sales","Profit","Quantity","Discount"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
## Insights

# - Sales and Profit show a positive correlation.
# - Discount has a negative correlation with Profit.
# - Quantity has a weak correlation with Sales and Profit.
# - Higher discounts generally reduce profitability.
```

## Sales Outlier Detection


```python
# Sales Outlier Detection

#A box plot is used to detect the distribution of Sales and identify outliers.
#Outliers represent unusually high or low sales values.

plt.figure(figsize=(6,4))

plt.boxplot(Data["Sales"])

plt.title("Sales Boxplot")

plt.ylabel("Sales")

plt.show()
```

## BUSINESS Dashboard

```python
# Business Dashboard

#This dashboard provides a quick overview of business performance using multiple visualizations.

#The dashboard includes:

#- Sales by Category
#- Profit by Category
#- Sales Distribution
#- Sales vs Profit

plt.figure(figsize=(16,10))

# -------------------------------
# Chart 1 : Sales by Category
# -------------------------------

plt.subplot(2,2,1)

category_sales = Data.groupby("Category")["Sales"].sum()

plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category")

plt.xlabel("Category")
plt.ylabel("Sales")


# -------------------------------
# Chart 2 : Profit by Category
# -------------------------------

plt.subplot(2,2,2)

category_profit = Data.groupby("Category")["Profit"].sum()

plt.bar(category_profit.index, category_profit.values,color="orange")

plt.title("Profit by Category")

plt.xlabel("Category")
plt.ylabel("Profit")


# -------------------------------
# Chart 3 : Sales Distribution
# -------------------------------

plt.subplot(2,2,3)

plt.hist(Data["Sales"],bins=30,color="green")

plt.title("Sales Distribution")

plt.xlabel("Sales")
plt.ylabel("Frequency")


# -------------------------------
# Chart 4 : Sales vs Profit
# -------------------------------

plt.subplot(2,2,4)

plt.scatter(Data["Sales"],Data["Profit"],color="red")

plt.title("Sales vs Profit")

plt.xlabel("Sales")
plt.ylabel("Profit")


plt.tight_layout()

plt.show()
    
```

```python
## Insights

# - Technology contributes the highest sales.
# - Profit varies significantly across categories.
# - Most sales values are concentrated at lower ranges.
# - Higher sales do not always generate higher profit.
# - A few high-value orders act as outliers.
# - The dashboard provides a quick overview of business performance.
```
